```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import (
    LabelEncoder,
    PolynomialFeatures,
    StandardScaler,
    MinMaxScaler
)

from sklearn.feature_selection import (
    VarianceThreshold,
    SelectKBest,
    f_regression
)

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("ai_chip_market.csv")

print(df.head())
```

```python
df.describe()
```

```python
print(df[["vendor", "chip_name"]].head())
df_fe = df.copy()

encoder = LabelEncoder()

df_fe["vendor"] = encoder.fit_transform(df_fe["vendor"])
df_fe["chip_name"] = encoder.fit_transform(df_fe["chip_name"])

print(df_fe[["vendor", "chip_name"]].head())
```

```python
print(df_fe[[
    "estimated_revenue_usd_m",
    "estimated_shipments_units"
]].head())

df_fe["Revenue_per_Shipment"] = (
    df_fe["estimated_revenue_usd_m"] /
    df_fe["estimated_shipments_units"]
)

print(df_fe[[
    "estimated_revenue_usd_m",
    "estimated_shipments_units",
    "Revenue_per_Shipment"
]].head())
```

```python
print(df_fe[["memory_gb"]].head(10))
# Binning by memory

df_fe["Memory_Category"] = pd.cut(
    df_fe["memory_gb"],
    bins=[0, 16, 32, 64, 200],
    labels=[
        "Low Memory",
        "Medium Memory",
        "High Memory",
        "Extreme Memory"
    ]
)

print(df_fe[["memory_gb", "Memory_Category"]].head(10))
```

```python
print(df_fe["Memory_Category"].value_counts())

df_fe["Memory_Category"].value_counts().sort_index().plot(
    kind="bar",
    color="skyblue"
)

plt.title("GPU Memory Categories")
plt.xlabel("Memory Category")
plt.ylabel("Count")
plt.show()
```

```python
#display dataset before and after applying feature engineering techniques
print("Before Feature Engineering:")
print(df.head())
print("\nAfter Feature Engineering:")
print(df_fe.head())
```

# Feature Engineering
### Label Encoding: vendor and chip_name columns are label encoded to convert categorical data into numerical format.
### New Feature Creation: A new feature 'Revenue_per_Shipment' is created by dividing 'estimated_revenue_usd_m' by 'estimated_shipments_units'.
### Binning: The 'memory_gb' column is binned into categories: Low Memory, Medium Memory, High Memory, and Extreme Memory, creating a new categorical feature 'Memory_Category'.

```python
target = "estimated_revenue_usd_m"
import seaborn as sns
import matplotlib.pyplot as plt

numeric_df = df_fe.select_dtypes(include="number")

corr = numeric_df.corr()

plt.figure(figsize=(10,7))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show()

print(corr[target].sort_values(ascending=False))
```

```python
print(df_fe.isna().sum())
df_fe = df_fe.dropna(subset=["Revenue_per_Shipment"])
print(df_fe["Revenue_per_Shipment"].isna().sum())
print(df_fe["Revenue_per_Shipment"].describe())
```

```python

X = df_fe.select_dtypes(include="number").drop(columns=[target])


y = df_fe[target]

selector = SelectKBest(score_func=f_regression, k=5)

selector.fit(X, y)

selected_features = X.columns[selector.get_support()]

print("Selected Features:")
print(selected_features)
```

```python
selector = VarianceThreshold(threshold=0.01)

X_var = selector.fit_transform(X)

selected_features = X.columns[selector.get_support()]

print(selected_features)
```

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LinearRegression

# Features
X = df_fe.select_dtypes(include="number").drop(columns=[target])

# Target
y = df_fe[target]

# Base model
model = LinearRegression()

# RFE
rfe = RFE(
    estimator=model,
    n_features_to_select=5
)

rfe.fit(X, y)

selected_features = X.columns[rfe.support_]

print("Selected Features:")
print(selected_features)

print("\nFeature Rankings:")
ranking = pd.DataFrame({
    "Feature": X.columns,
    "Rank": rfe.ranking_
})

print(ranking.sort_values("Rank"))
```

```python
from sklearn.ensemble import RandomForestRegressor

# Features
X = df_fe.select_dtypes(include="number").drop(columns=[target])

# Target
y = df_fe[target]

# Train model
rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf.fit(X, y)

importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": rf.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

print(importance)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.barh(
    importance["Feature"],
    importance["Importance"]
)

plt.title("Tree-Based Feature Importance")

plt.xlabel("Importance")

plt.gca().invert_yaxis()

plt.show()
```

# Feature Selection
### Correlation Analysis: A heatmap is generated to visualize the correlation between numerical features and the target variable 'estimated_revenue_usd_m'. Features with high correlation to the target are identified.
### SelectKBest: The SelectKBest method is used with the f_regression scoring function to select the top 5 features that have the strongest relationship with the target variable: 'year', 'chip_name', 'vendor', 'fp16_tflops', 'estimated_shipments_units'
### Variance Threshold: Features with low variance (below a threshold of 0.01) are removed, as they are unlikely to be informative for the model. The features selected are: 'year', 'chip_name', 'vendor', 'memory_gb', 'fp16_tflops', 'tdp_watts', 'estimated_shipments_units', 'estimated_asp_usd', 'Revenue_per_Shipment'
### Recursive Feature Elimination (RFE): RFE is applied using a linear regression model to recursively eliminate less important features, selecting the top 5 features based on their contribution to the model's performance: 'year', 'chip_name', 'vendor', 'memory_gb', 'fp16_tflops'
### Tree-Based Feature Importance: A Random Forest Regressor is trained, and the importance of each feature is calculated based on how much they contribute to reducing the impurity in the trees. The features are ranked by their importance scores, and a bar chart is generated to visualize the importance of each feature: 'estimated_shipments_units' and 'year' are the two most important features according to the Random Forest model.

```python

cols = [
    "memory_gb",
    "fp16_tflops",
    "tdp_watts",
    "estimated_shipments_units",
    "estimated_asp_usd"
]

standard = StandardScaler()

scaled = standard.fit_transform(df_fe[cols])

standard_df = pd.DataFrame(
    scaled,
    columns=cols
)

print(standard_df.head())
```

```python

minmax = MinMaxScaler()

scaled = minmax.fit_transform(df_fe[cols])

minmax_df = pd.DataFrame(
    scaled,
    columns=cols
)

print(minmax_df.head())
```

```python
from sklearn.preprocessing import RobustScaler

cols = [
    "memory_gb",
    "fp16_tflops",
    "tdp_watts",
    "estimated_shipments_units",
    "estimated_asp_usd"
]

robust = RobustScaler()

robust_scaled = robust.fit_transform(df_fe[cols])

robust_df = pd.DataFrame(
    robust_scaled,
    columns=cols
)

print(robust_df.head())
```

```python
from sklearn.preprocessing import MaxAbsScaler

cols = [
    "memory_gb",
    "fp16_tflops",
    "tdp_watts",
    "estimated_shipments_units",
    "estimated_asp_usd"
]

maxabs = MaxAbsScaler()

maxabs_scaled = maxabs.fit_transform(df_fe[cols])

maxabs_df = pd.DataFrame(
    maxabs_scaled,
    columns=cols
)

print(maxabs_df.head())
```

```python
comparison = pd.concat(
    [
        df_fe[cols].head(),
        standard_df.head(),
        minmax_df.head(),
        robust_df.head(),
        maxabs_df.head()
    ],
    axis=1,
    keys=[
        "Original",
        "Standard Scaling",
        "Min-Max Scaling",
        "RobustScaler",
        "MaxAbsScaler"
    ]
)

comparison
```
