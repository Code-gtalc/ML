```python
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import f_regression

X = df_fe.select_dtypes(include=np.number)

X = X.drop(columns=[target])

y = df_fe[target]

selector = SelectKBest(
    score_func=f_regression,
    k=5
)

selector.fit(X,y)

selected = X.columns[
    selector.get_support()
]

print(selected)
```

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LinearRegression

X = df_fe.select_dtypes(include=np.number)

X = X.drop(columns=[target])

y = df_fe[target]

model = LinearRegression()

rfe = RFE(
    estimator=model,
    n_features_to_select=5
)

rfe.fit(X,y)

selected = X.columns[rfe.support_]

print(selected)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    random_state=42
)

rf.fit(X,y)

importance = pd.DataFrame({
    "Feature":X.columns,
    "Importance":rf.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

print(importance)
```

```python
plt.figure(figsize=(8,5))

plt.barh(
    importance["Feature"],
    importance["Importance"]
)

plt.gca().invert_yaxis()

plt.title("Feature Importance")

plt.show()
```

```python
cols = [
    "positive",
    "negative",
    "userscore",
    "average_forever",
    "price",
    "initialprice",
    "ccu"
]
```

```python
from sklearn.preprocessing import StandardScaler

standard = StandardScaler()

standard_scaled = standard.fit_transform(df_fe[cols])

standard_df = pd.DataFrame(
    standard_scaled,
    columns=cols
)

standard_df.head()
```

```python
fig, ax = plt.subplots(1,2, figsize=(12,5))

sns.histplot(
    df_fe["price"],
    ax=ax[0],
    kde=True
)

ax[0].set_title("Original Price")

sns.histplot(
    standard_df["price"],
    ax=ax[1],
    kde=True
)

ax[1].set_title("Standard Scaled Price")

plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler

minmax = MinMaxScaler()

minmax_scaled = minmax.fit_transform(
    df_fe[cols]
)

minmax_df = pd.DataFrame(
    minmax_scaled,
    columns=cols
)

minmax_df.head()
```

```python
from sklearn.preprocessing import RobustScaler

robust = RobustScaler()

robust_scaled = robust.fit_transform(df_fe[cols])

robust_df = pd.DataFrame(
    robust_scaled,
    columns=cols
)

robust_df.head()
```

```python
fig, ax = plt.subplots(1,2, figsize=(12,5))

sns.histplot(
    df_fe["price"],
    ax=ax[0],
    kde=True
)

ax[0].set_title("Original Price")

sns.histplot(
    robust_df["price"],
    ax=ax[1],
    kde=True
)

ax[1].set_title("Robust Scaled Price")

plt.show()
```

```python
from sklearn.preprocessing import MaxAbsScaler

maxabs = MaxAbsScaler()

maxabs_scaled = maxabs.fit_transform(
    df_fe[cols]
)

maxabs_df = pd.DataFrame(
    maxabs_scaled,
    columns=cols
)

maxabs_df.head()
```

```python
comparison = pd.concat(
    [
        df_fe[cols].head(),
        standard_df.head(),
        robust_df.head()
    ],
    axis=1,
    keys=[
        "Original",
        "StandardScaler",
        "MinMaxScaler"
        "RobustScaler"
        "MaxAbsScaler"
    ]
)

comparison
```
