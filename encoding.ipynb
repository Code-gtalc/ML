```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df=pd.read_excel("Dataset.xlsx")

```

```python
#Label encoding in gender column
le=LabelEncoder()
le.fit(df["Gender"])
df["Gender"]=le.transform(df["Gender"])
print(df)
```

```python
#Ordinal encoding in Education_Level column

education_order = ["High School", "Bachelor's", "Master's", "PhD"]
df["Education_Level"] = pd.Categorical(df["Education_Level"], categories=education_order, ordered=True)
df["Education_Level"] = df["Education_Level"].cat.codes
print(df)
```

```python
#Frequency encoding on City column

city_freq = df["City"].value_counts()
df["City"] = df["City"].map(city_freq)
print(df)
```

```python
#One-hot encoding on Department column

df = pd.get_dummies(df, columns=["Department"], prefix="Dept")
print(df)
```
