```python
# Standardize Salary Columns

import re

def clean_salary(value):

    # Handle missing values
    if pd.isna(value):
        return 0.0

    value = str(value).strip().lower()

    # Remove commas and extra spaces
    value = value.replace(",", "")
    value = re.sub(r"\s+", "", value)

    # Handle empty/invalid responses
    invalid = {
        "", "na", "n/a", "none", "null",
        "nil", "-", "--", "unknown"
    }

    if value in invalid:
        return 0.0

    # Handle ranges (e.g., 10-12 LPA, 10to12LPA)
    range_match = re.findall(r"\d+\.?\d*", value)

    if ("-" in value or "to" in value) and len(range_match) >= 2:
        number = (float(range_match[0]) + float(range_match[1])) / 2

        if "cr" in value or "crore" in value:
            return number * 100

        elif "lpa" in value or "lakh" in value or "lac" in value:
            return number

        else:
            return number

    # Extract first numeric value
    match = re.search(r"\d+\.?\d*", value)

    if not match:
        return 0.0

    number = float(match.group())

    # -------- UNIT CONVERSIONS -------- #

    # Crore -> LPA
    if "crore" in value or "cr" in value:
        return number * 100

    # Lakh/LPA -> LPA
    elif "lpa" in value or "lakh" in value or "lac" in value:
        return number

    # Rupees -> LPA
    elif number >= 100000:
        return number / 100000

    # Already numeric LPA
    else:
        return number


# Apply to all salary columns
for col in salary_cols:
    df[col] = df[col].apply(clean_salary)

# Verify results
df[salary_cols]
```

```python
import datetime

def clean_percentage(value):

    if isinstance(value, (datetime.datetime, datetime.date)):
        return 0.0

    if pd.isna(value):
        return 0.0

    value = str(value).lower().strip()

    match = re.search(r'\d+\.?\d*', value)

    if not match:
        return 0.0

    num = float(match.group())

    # Excel decimal percentage
    if 0 < num <= 1:
        num *= 100

    return max(0,min(num,100))


for col in percentage_cols:
    df[col] = df[col].apply(clean_percentage)
df[percentage_cols]
```

```python
import re
import datetime


def percentage_to_gpa4(p):

    if p >= 80:
        return 4.0
    elif p >= 70:
        return 3.6
    elif p >= 60:
        return 3.2
    elif p >= 55:
        return 2.8
    elif p >= 50:
        return 2.4
    elif p >= 45:
        return 2.0
    elif p >= 40:
        return 1.6
    else:
        return 0.0


def clean_gpa(value):

    # Datetime accidentally entered
    if isinstance(value, (datetime.datetime, datetime.date)):
        return 0.0

    if pd.isna(value):
        return 0.0

    value = str(value).lower().strip()

    # Replace dash used as decimal separator
    value = re.sub(r'(?<=\d)-(?=\d)', '.', value)

    # Handle x/y formats
    frac = re.search(r'(\d+\.?\d*)/(\d+\.?\d*)', value)

    if frac:

        num = float(frac.group(1))
        den = float(frac.group(2))

        if den == 4:
            return min(num,4.0)

        elif den == 10:
            return percentage_to_gpa4(num*10)

        elif den == 100:
            return percentage_to_gpa4(num)

    match = re.search(r'\d+\.?\d*', value)

    if not match:
        return 0.0

    num = float(match.group())

    # Excel percentage
    if 0 < num <= 1:
        return percentage_to_gpa4(num*100)

    # Percentage
    if "%" in value or num >= 10:
        return percentage_to_gpa4(num)

    # Already 4-point GPA
    if num <= 4:
        return num

    # 10-point GPA
    return percentage_to_gpa4(num*10)
gpa_col=["your GPA of last semester"]
for col in gpa_col:
    df["your GPA of last semester"] = df["your GPA of last semester"].apply(clean_gpa)

df["your GPA of last semester"]
```

```python
threshold = len(df.columns)//2

df = df.dropna(thresh=threshold)

print(df.shape)
```

```python
# Mean / Median Imputation

numeric_cols = df.select_dtypes(include="number").columns

for col in numeric_cols:

    # Skip columns that have no missing values
    if df[col].isna().sum() == 0:
        continue

    # Use median for skewed distributions
    if abs(df[col].skew()) > 1:
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna(df[col].mean())
```

```python
# Ensure cleaned numeric columns are numeric dtype

numeric_conversion_cols = salary_cols + percentage_cols + [
    "your GPA of last semester"
]

for col in numeric_conversion_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
# Mode Imputation

categorical_cols = df.select_dtypes(include="number").columns

for col in categorical_cols:

    if df[col].isna().sum() == 0:
        continue

    mode = df[col].mode()

    if not mode.empty:
        df[col] = df[col].fillna(mode.iloc[0])
```

```python
def cap_outliers(series):

    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR

    return series.clip(lower, upper)
```

```python
outlier_cols = [
    "What are your package expectations (LPA)",
    "Your maximum attendance % till last semester",
    "your GPA of last semester"
]

for col in outlier_cols:
    df[col] = cap_outliers(df[col])

print("Outliers treated successfully.")
```
